# Chapter 7 — ConfigMaps and Secrets

This chapter explains how Kubernetes manages:

* application configuration,
* environment variables,
* passwords,
* API keys,
* database credentials,
* runtime settings.

This is where Kubernetes becomes a true:

> production platform.

Because real applications need more than:

* compute,
* networking,
* storage.

They also need:

> dynamic configuration management.

---

# City Analogy Expansion

| Kubernetes            | City Analogy                     |
| --------------------- | -------------------------------- |
| ConfigMap             | Public instruction documents     |
| Secret                | Confidential government vault    |
| Environment Variables | Instructions given to businesses |
| Mounted Config Files  | Notice boards/documents          |
| Pod                   | Apartment building               |
| Application           | Business running inside building |

---

# The Big Problem

Applications need configuration.

Examples:

* database URL,
* API endpoint,
* feature flags,
* environment mode,
* passwords,
* tokens,
* certificates.

---

# BAD Old Approach

Developers often hardcoded config inside app code.

Example:

```javascript id="4c7kaf"
db_password = "mypassword123"
```

Very dangerous.

---

# Problems With Hardcoding

---

## 1. Security Risk

Passwords inside source code.

---

## 2. Difficult Environment Changes

Dev/staging/prod all need different configs.

---

## 3. Redeployment Needed

Tiny config change requires rebuilding app.

---

# Kubernetes Philosophy

Kubernetes strongly separates:

> application code
> FROM
> application configuration.

This is a huge production engineering principle.

---

# City Analogy

Businesses should NOT permanently print:

* passwords,
* city policies,
* utility settings,
  inside their furniture.

Instead:

* city administration distributes updated instructions dynamically.

---

# Part 1 — What Is a ConfigMap?

ConfigMap stores:

> non-sensitive configuration data.

Examples:

* app mode,
* URLs,
* feature toggles,
* port numbers,
* configuration files.

---

# City Analogy

ConfigMap = public city instruction documents.

Examples:

* traffic timings,
* public regulations,
* operational guidelines.

Not confidential.

---

# Example ConfigMap

```yaml id="8e0x4u"
APP_MODE=production
LOG_LEVEL=debug
API_URL=payments.company.com
```

---

# Important Understanding

ConfigMaps are NOT for secrets.

They are:

* plain configuration,
* non-sensitive information.

---

# Part 2 — How Pods Use ConfigMaps

Pods can consume ConfigMaps in multiple ways.

---

# Method 1 — Environment Variables

Most common approach.

---

# Example

Inside pod:

```text id="v8f2s6"
APP_MODE=production
```

Application reads variable normally.

---

# City Analogy

Business receives operational instructions from city administration.

---

# Method 2 — Mounted Files

ConfigMap can appear as files inside container.

---

# Example

```text id="l4l5vz"
/config/app-config.yaml
```

---

# Why Mounted Files Are Useful

Some applications expect:

* configuration files,
  NOT environment variables.

Example:

* NGINX,
* Apache,
* many databases.

---

# City Analogy

Instead of verbal instructions:

* official documents posted on office wall.

---

# Part 3 — What Is a Secret?

Secrets store:

> sensitive information.

Examples:

* passwords,
* API keys,
* OAuth tokens,
* TLS certificates,
* private keys.

---

# City Analogy

Secret = confidential government vault.

Only authorized businesses can access.

---

# Important Difference

---

## ConfigMap

Public instructions.

---

## Secret

Confidential information.

---

# Example Secret

```text id="yn1x8u"
DB_PASSWORD
JWT_SECRET
API_TOKEN
```

---

# Internal Reality

Secrets are base64-encoded in Kubernetes.

IMPORTANT:

* encoding ≠ encryption.

This confuses many beginners.

---

# Production Security

Real production clusters often additionally use:

* encryption at rest,
* external secret managers,
* vault systems.

Examples:

* HashiCorp Vault
* Amazon Web Services Secrets Manager

---

# Part 4 — Why Secrets Exist Separately

Kubernetes intentionally separates:

* normal config,
* sensitive config.

Because:

* permissions,
* auditing,
* security handling differ.

---

# City Analogy

Public city guidelines should NOT be stored with:

* nuclear launch codes,
* bank vault keys,
* police access credentials.

---

# Part 5 — Injecting Secrets Into Pods

Secrets can also be:

* environment variables,
* mounted files.

---

# Example

```text id="0p1k2v"
DB_PASSWORD=******
```

Application reads normally.

---

# Important Production Practice

Applications should NEVER:

* hardcode secrets,
* commit secrets into Git repositories.

Very common beginner mistake.

---

# Part 6 — Environment Variables

Kubernetes commonly injects config through:

> environment variables.

---

# Example

Application receives:

```text id="n2j1b7"
DATABASE_HOST
DATABASE_PORT
REDIS_URL
```

This allows:

* same application image,
* different runtime environments.

---

# Huge Production Benefit

One Docker image can run in:

* dev,
* testing,
* staging,
* production.

Only configuration changes.

---

# City Analogy

Same business franchise:

* different operating rules in different cities.

Business structure unchanged.

---

# Part 7 — Dynamic Configuration

Suppose:

* logging level changes,
* feature toggle enabled,
* API endpoint updated.

Instead of rebuilding app:

* update ConfigMap.

---

# Why This Matters

Without configuration separation:

* tiny config changes require redeployment.

Very inefficient.

---

# Kubernetes Makes Apps More Portable

Portable applications are:

* environment-independent,
* infrastructure-independent.

Configuration injection enables portability.

---

# Part 8 — Mounted Configuration Files

Very important in enterprise systems.

Some applications expect:

```text id="yk6j0t"
nginx.conf
app.properties
database.yaml
```

Kubernetes can mount ConfigMaps directly as files.

---

# Example Internal Flow

```text id="9hpk8u"
ConfigMap
   ↓ mounted as
/config/settings.yaml
```

Container reads file normally.

---

# City Analogy

Official policy documents physically delivered to office.

---

# Part 9 — Secret Security Concerns

Even with Kubernetes Secrets:

* security is NOT automatically solved.

---

# Common Problems

---

## 1. Base64 Misunderstanding

Many beginners think:

```text id="y12b9q"
base64 = encrypted
```

Wrong.

Base64 is easily reversible.

---

## 2. Secrets in Logs

Applications accidentally print secrets.

---

## 3. Over-Permissioned Access

Too many pods can access secrets.

---

# Real Production Security

Advanced setups use:

* RBAC,
* encryption,
* secret rotation,
* external vault systems.

---

# Part 10 — Why Config Management Is Critical

Modern systems are highly dynamic.

Applications constantly need:

* changing endpoints,
* rotating credentials,
* feature toggles,
* environment-specific settings.

Kubernetes centralizes this management.

---

# Real-World Example

Suppose:

* Spotify deploys same service globally.

Different regions need:

* different database endpoints,
* different API keys,
* different feature flags.

ConfigMaps and Secrets allow:

* same application image,
* different runtime behavior.

---

# Internal Flow Example

---

# Step 1 — Create ConfigMap

Contains:

```text id="g9w6jq"
APP_MODE=production
```

---

# Step 2 — Pod References ConfigMap

Kubernetes injects config.

---

# Step 3 — Application Reads Config

Application starts with proper settings.

---

# Similar Secret Flow

```text id="t9zg49"
Secret
   ↓
Pod
   ↓
Environment Variable / Mounted File
   ↓
Application Uses Credential
```

---

# Important Beginner Mistakes

---

## Mistake 1

Putting passwords inside ConfigMaps.

Wrong.
Use Secrets.

---

## Mistake 2

Hardcoding credentials in images.

Very dangerous.

---

## Mistake 3

Confusing Secret encoding with encryption.

Base64 is NOT secure encryption.

---

## Mistake 4

Rebuilding images for config changes.

Configuration should be externalized.

---

# Kubernetes Core Philosophy Again

Kubernetes separates:

* application code,
* configuration,
* secrets,
* infrastructure.

This separation creates:

* portability,
* scalability,
* operational flexibility.

---
