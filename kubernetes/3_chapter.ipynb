# The Biggest Kubernetes Secret

Kubernetes is fundamentally:

> A giant continuous reconciliation system.

Meaning:

1. You declare desired state.
2. Kubernetes continuously compares:

   * current state
   * desired state
3. Kubernetes fixes differences automatically.

---

# City Analogy

City law says:

> “This city must always have 100 hospitals.”

Government continuously checks:

* are there 100?
* did one collapse?
* did population increase?
* is traffic overloaded?

Government continuously repairs imbalance.

That continuous correction mechanism is Kubernetes.

---

# High-Level Kubernetes Architecture

Kubernetes has two major parts:

---

# 1. Control Plane (Brain)

Makes decisions.

---

# 2. Worker Nodes (Execution Layer)

Actually run applications.

you are talking to:

> API Server.

NOT directly to nodes.

---

# City Analogy

API Server = city reception office.

Every request enters here:

* building permits,
* complaints,
* population updates,
* repair requests.

Nothing bypasses reception office.

---

# Example Flow

You say:

```yaml
I want 3 pods
```

API Server:

* validates request,
* stores it,
* informs other components.

---

# Kubernetes Is API-Driven

This is EXTREMELY important.

Kubernetes internally works like:

* a distributed API system.

Everything:

* reads from API,
* writes to API,
* watches API.

---

# Part 3 — etcd

etcd is:

> Kubernetes’ database.

It stores:

* cluster state,
* pod information,
* configurations,
* secrets,
* deployments,
* everything important.

---

# City Analogy

etcd = central city registry office.

Stores records like:

* land ownership,
* building maps,
* citizen records,
* laws,
* city status.

If registry disappears:

* city loses memory.

---

# Critical Fact

etcd is the MOST critical Kubernetes component.

If etcd is corrupted:

* cluster becomes unusable.

---

# Example

Suppose cluster contains:

* 500 nodes,
* 10,000 pods.

etcd stores:

* what exists,
* where,
* desired configuration.

---

# Part 4 — Scheduler

Scheduler decides:

> Which node should run a pod.

This is one of Kubernetes’ smartest systems.

---

# Scheduler Checks

When placing pod:

* CPU availability,
* memory,
* node health,
* affinity rules,
* taints/tolerations,
* workload balance.

---

# City Analogy

Scheduler = urban planning department.

When new apartment building must be constructed:

* choose colony with enough space,
* electricity,
* water,
* traffic capacity.

---

# Example

Suppose:

* Node A has 90% CPU used,
* Node B has 20% CPU used.

Scheduler likely chooses:

* Node B.

---

# Important Understanding

You usually do NOT manually choose node.

Kubernetes intelligently schedules workloads.

---

# Part 5 — Controller Manager

Controller Manager is:

> The continuous repair and monitoring system.

This is Kubernetes’ “self-healing” brain.

---

# Controllers Continuously Watch

Example:
Desired:

```text id="5a44pc"
5 pods
```

Actual:

```text id="94oc4v"
4 pods
```

Controller immediately notices mismatch.

Then:

* creates new pod.

---

# City Analogy

Repair department constantly checks:

* damaged buildings,
* missing hospitals,
* overloaded roads.

Then automatically repairs city.

---

# This Is the Heart of Kubernetes

Controllers continuously perform:

```text id="8j7ozf"
Observe → Compare → Correct
```

Again and again forever.

---

# Important Kubernetes Philosophy

Kubernetes is NOT:

* command execution system.

It is:

> a desired-state reconciliation engine.

This idea is EVERYTHING.

---

# Part 6 — Kubelet

Kubelet runs on every node.

It is:

> The node agent.

Kubelet communicates with control plane.

---

# Responsibilities

Kubelet:

* receives pod instructions,
* starts containers,
* monitors health,
* reports status.

---

# City Analogy

Kubelet = colony supervisor.

Central government says:

> “Build 3 apartment buildings.”

Local colony supervisor ensures:

* buildings actually constructed,
* maintained properly.

---

# Internal Flow

1. API Server stores desired pod
2. Scheduler selects node
3. Kubelet on node sees assignment
4. Kubelet starts containers

---

# Part 7 — Container Runtime

Container runtime actually runs containers.

Examples:

* containerd,
* CRI-O.

Earlier:

* Docker was commonly used.

---

# City Analogy

Container runtime = construction workers.

Government plans buildings.

Workers physically build them.

---

# Part 8 — Kube Proxy

Kube Proxy handles:

* networking,
* routing,
* service traffic.

---

# City Analogy

Traffic police + road routing system.

Ensures vehicles reach correct buildings.

---

# Complete Internal Flow of Pod Creation

This is VERY important.

Let’s follow entire lifecycle.

---

# Step 1 — User Sends Request

You run:

```bash
kubectl apply -f app.yaml
```

Request goes to:

> API Server.

---

# City Analogy

Business submits construction request to city office.

---

# Step 2 — API Server Validates

Checks:

* syntax,
* permissions,
* correctness.

Then stores desired state in:

> etcd.

---

# Step 3 — Scheduler Assigns Node

Scheduler finds best node.

---

# City Analogy

Urban planning department selects colony.

---

# Step 4 — Kubelet Receives Instruction

Node’s kubelet notices:

> “New pod assigned to me.”

---

# Step 5 — Container Runtime Starts Containers

Containers begin running.

---

# Step 6 — Status Updates Return

Kubelet reports:

* running,
* failed,
* restarting,
* healthy.

Back to control plane.

---
