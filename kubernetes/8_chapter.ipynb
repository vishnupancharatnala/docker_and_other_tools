# The Big Problem

In Kubernetes:

* many pods share same nodes,
* nodes have limited resources.

Suppose one pod consumes:

* all CPU,
* all memory.

Other applications fail.

Kubernetes needs:

> intelligent resource management.

---

# City Analogy

A city has limited:

* electricity,
* water,
* road capacity.

One factory should NOT consume entire city power supply.

City administration controls resource allocation.

---

# Part 1 — CPU and Memory Basics

Kubernetes mainly manages:

* CPU,
* Memory (RAM).

These are the most critical compute resources.

---

# CPU

CPU represents:

> processing power.

Examples:

* request handling,
* computations,
* encoding,
* API processing.

---

# Memory (RAM)

Memory stores:

* running application state,
* caches,
* temporary runtime data.

---

# Important Difference

---

## CPU can be throttled

Application slowed down.

---

## Memory cannot be safely throttled

If memory exhausted:

> container may be killed.

Very important production reality.

---

# City Analogy

---

## CPU = electricity consumption

Can reduce power temporarily.

---

## Memory = physical water tank capacity

If tank overflows:

* system fails.

---

# Part 2 — Requests and Limits

This is one of Kubernetes’ most important concepts.

---

# Requests

Request =

> minimum guaranteed resources.

Example:

```yaml id="g5k4m7"
cpu: 500m
memory: 512Mi
```

Scheduler uses requests when placing pods.

---

# City Analogy

Business says:

> “I need at least 500 units of electricity and 512 liters water.”

City reserves that amount.

---

# Limits

Limit =

> maximum allowed usage.

Example:

```yaml id="f8q0vx"
cpu: 1
memory: 1Gi
```

Application cannot exceed limit safely.

---

# Why Limits Exist

Without limits:

* buggy application could consume entire node.

Very dangerous.

---

# City Analogy

City law:

> “Factory cannot exceed certain electricity usage.”

Prevents city-wide outages.

---

# Part 3 — Scheduler and Resource Placement

Scheduler uses:

* requests,
* node capacity,
* constraints.

To decide:

> where pods should run.

---

# Example

Node A:

```text id="2c6y8q"
90% CPU used
```

Node B:

```text id="2k2o4e"
20% CPU used
```

Scheduler prefers:

* Node B.

---

# Kubernetes Goal

Efficient resource utilization while preventing overload.

---

# City Analogy

Urban planning department places:

* new buildings,
* factories,
* apartments,

based on available:

* electricity,
* water,
* road capacity.

# Part 4 — What Happens During Resource Exhaustion?

Very important production topic.

---

# CPU Exhaustion

If pod exceeds CPU limit:

* Kubernetes throttles CPU.

Application slows down.

---

# Memory Exhaustion

If pod exceeds memory limit:

> container may be OOMKilled.

OOM =

```text id="m7cn4e"
Out Of Memory
```

Container terminated.

---

# Why Memory Is Dangerous

Memory cannot be compressed safely.

Kernel eventually kills processes.

---

# City Analogy

---

## CPU overload

Temporary electricity reduction.

---

## Memory overload

Water tank explosion forcing shutdown.

---

# Part 5 — Horizontal Pod Autoscaler (HPA)

One of Kubernetes’ coolest features.

HPA automatically:

> increases or decreases pod count.

Based on:

* CPU,
* memory,
* metrics,
* traffic.

---

# Example

Suppose:

* average CPU > 80%.

HPA scales:

```text id="k3dz8z"
5 pods → 10 pods
```

Traffic decreases:

```text id="35u1mv"
10 pods → 3 pods
```

Automatic.

---

# City Analogy

Festival season arrives.

City automatically:

* opens more transport,
* builds temporary facilities,
* expands services.

Festival ends:

* extra infrastructure removed.

---

# Why HPA Is Revolutionary

Traditional systems often required:

* manual scaling,
* operations teams,
* downtime planning.

Kubernetes scales dynamically.

---

# Part 6 — Vertical Pod Autoscaler (VPA)

Instead of changing pod count:
VPA changes:

> pod resource sizes.

Example:

```text id="bj8n5g"
CPU: 500m → 2 CPU
```

---

# Difference

---

## HPA

More pods.

---

## VPA

Bigger pods.

---

# City Analogy

---

## HPA

Construct more apartment buildings.

---

## VPA

Expand existing building size.

---

# Production Reality

HPA more commonly used for:

* stateless applications.

VPA often trickier because:

* restarting pods may be needed.

---

# Part 7 — Cluster Autoscaler

Suppose:

* all nodes become full.

Even HPA cannot create more pods.

Kubernetes can automatically:

> add more nodes.

This is Cluster Autoscaler.

---

# Example

Traffic spike:

```text id="50lw5e"
Need 100 more pods
```

No node capacity available.

Cloud provider automatically creates:

* new VMs/nodes.

---

# City Analogy

City becomes overcrowded.

Government builds:

* entirely new colonies.

---

# This Enables Massive Cloud Scalability

Platforms like:

* Amazon Web Services
* Google Cloud
* Microsoft Azure

can dynamically scale infrastructure globally.

---

# Part 8 — Resource Quotas

Large clusters often host:

* multiple teams,
* multiple projects.

Need fairness.

---

# ResourceQuota

Limits namespace resource usage.

Example:

```text id="v00t8o"
Team A cannot exceed:
- 50 CPUs
- 100GB memory
```

---

# City Analogy

City laws prevent one company from consuming:

* all electricity,
* all water,
* all land.

# Part 11 — Burstable vs Guaranteed Pods

Kubernetes classifies pods by resource guarantees.

---

# Guaranteed

Requests = Limits.

Highest stability.

---

# Burstable

Can temporarily use extra resources.

---

# BestEffort

No guarantees.

Lowest priority.

---

# During Node Pressure

Kubernetes may kill:

1. BestEffort first
2. Burstable next
3. Guaranteed last

---

# City Analogy

Emergency power shortages:

* low-priority consumers disconnected first.

Critical hospitals protected.
