> a controlled automated process.

# Part 1 — Rolling Updates

This is Kubernetes’ default deployment strategy.

Instead of replacing everything at once:

* new pods gradually added,
* old pods gradually removed.

---

# Example

Initially:

```text id="v7l4x0"
10 pods → version v1
```

Deployment starts update:

```text id="r5s6g2"
8 pods → v1
2 pods → v2
```

Then:

```text id="k0m2v7"
5 pods → v1
5 pods → v2
```

Eventually:

```text id="z6c1t9"
10 pods → v2
```

Application remains available entire time.

---

# City Analogy

Instead of demolishing all apartment buildings:

* renovate one building at a time,
* citizens continue living normally.

---

# Why Rolling Updates Are Powerful

They provide:

* zero downtime,
* safer deployments,
* gradual verification,
* automatic rollback possibility.

---

# Part 2 — Readiness Probes

Very important Kubernetes feature.

How does Kubernetes know:

> new pod is ready to receive traffic?

It uses:

> readiness probes.

---

# Example

Kubernetes checks:

```text id="w1n8e4"
GET /health
```

If healthy:

* traffic allowed.

If unhealthy:

* pod removed from service endpoints.

---

# City Analogy

Before opening renovated building:

* city inspectors verify:

  * electricity,
  * water,
  * safety.

Only then citizens allowed entry.

---

# Important Difference

---

## Pod Running

Container started.

---

## Pod Ready

Actually capable of serving users.

Huge difference.

---

# Part 3 — Liveness Probes

Liveness probes detect:

> stuck or unhealthy applications.

If liveness fails:

* Kubernetes restarts container.

---

# Example

Application deadlocked:

* process alive,
* but app non-functional.

Liveness probe detects issue.

---

# City Analogy

Building exists physically,
but:

* elevators broken,
* utilities failed,
* unusable internally.

City repairs/restarts operations.

---

# Kubernetes Self-Healing Deepens Here

Kubernetes continuously monitors:

* process health,
* application responsiveness,
* readiness state.

---

# Part 4 — Rollbacks

Suppose deployment fails.

Kubernetes can rollback quickly.

---

# Example

New version:

```text id="2m9g0z"
v2
```

Causes:

* crashes,
* high latency,
* errors.

Deployment rollback:

```text id="3f7d1q"
v1 restored
```

---

# City Analogy

New building design unsafe.

City immediately restores previous approved design.

---

# Why Rollbacks Matter

In production:

* failures are inevitable.

Fast rollback minimizes:

* downtime,
* business damage,
* customer impact.

---

# Part 5 — CI/CD Pipelines

CI/CD =

* Continuous Integration
* Continuous Deployment/Delivery

This automates software delivery.

---

# Typical Flow

```text id="4v8t2p"
Developer writes code
        ↓
Tests run automatically
        ↓
Container image built
        ↓
Image pushed to registry
        ↓
Kubernetes deployment updated
```

---

# Benefits

* instant rollback,
* safer releases,
* easier testing.

---

# Drawback

Requires:

* double infrastructure.

Expensive.

---

# City Analogy

# Part 8 — Canary Deployments

Canary release =

> expose small percentage of users to new version.

---

# Example

```text id="n5h8c0"
95% traffic → v1
5% traffic → v2
```

Monitor results.

If healthy:

* gradually increase traffic.

---

# Why “Canary”?

Historically miners used:

* canary birds.

If toxic gas present:

* canary died first.

Small controlled testing.

---

# City Analogy

New bridge opened first to:

* small controlled traffic.

Verify safety before city-wide opening.

---

# This Is Extremely Common

Large companies like:

* Netflix
* Google
* Uber

heavily use progressive rollout strategies.

---

# Part 9 — GitOps

Very modern Kubernetes operational model.

---

# Core Idea

Git repository becomes:

> source of truth.

Desired infrastructure stored in Git.

Automation continuously syncs cluster state.

---

# Example

Developer updates:

```yaml id="4x0j7p"
replicas: 10
```

# Part 10 — Immutable Infrastructure

Very important DevOps principle.

Instead of modifying running systems:

> replace them entirely.

---

# Kubernetes Follows This Strongly

New version deployment:

* create new pods,
* destroy old pods.

Not:

* manually SSH into servers,
* patch systems.

---

# City Analogy

* move citizens,
* demolish old structures.

Cleaner and more reliable.
