# Virtual Machines (VMs)

## Prerequisites / Basic Computer Architecture

A normal computer works like this:

```text
Applications
     |
Operating System (Windows/Linux/macOS)
     |
Hardware (CPU, RAM, Disk, Network)
```

* Applications use the Operating System (OS).
* The OS directly communicates with the hardware.

---

# What is a Virtual Machine?

Suppose your laptop has **Windows** installed, but you also want to use **Linux**.

Instead of buying another laptop, you can create a **Virtual Machine (VM)** inside your existing computer.

A VM is a **software-based computer** that behaves like a real computer.

The VM contains:

* Virtual CPU
* Virtual RAM
* Virtual Disk
* Guest Operating System (Linux, Windows, etc.)

The operating system running inside the VM is called the **Guest OS**.

The original operating system of your laptop is called the **Host OS**.

The Guest OS feels like it is running on a real physical machine.

---

# Hypervisor

A **Hypervisor** is software (or firmware) that creates and manages Virtual Machines.

Responsibilities of a Hypervisor:

* Creates VMs
* Allocates CPU, RAM, and storage
* Isolates VMs from each other
* Manages communication between VMs and hardware

Examples:

* VMware
* VirtualBox
* Hyper-V
* KVM

---

# Types of Hypervisors

## Type 1 Hypervisor (Bare Metal Hypervisor)

Architecture:

```text
Virtual Machines
      |
Hypervisor
      |
Hardware
```

Examples:

* VMware ESXi
* Microsoft Hyper-V Server
* Xen

### Features

* Runs directly on hardware
* No host operating system required
* Faster and more efficient
* More secure
* Used in data centers and cloud environments

### Usage

Used mainly in:

* Organizations
* Cloud computing
* Servers
* Data centers

---

## Type 2 Hypervisor (Hosted Hypervisor)

Architecture:



```text
Virtual Machines
      |
Hypervisor
      |
Host Operating System
      |
Hardware
```

Examples:

* Oracle VirtualBox
* VMware Workstation
* Parallels Desktop

### Features

* Runs as a normal application on an existing OS
* Easier to install and use
* Slightly slower than Type 1

### Usage

Used mainly for:

* Learning
* Testing
* Development
* Running another OS on personal laptops/desktops

---

# Advantages of Virtual Machines

* Run multiple operating systems on one computer
* Better resource utilization
* Isolation between environments
* Easy testing and experimentation
* Snapshots and backups are possible
* Useful for cloud computing

---

# Disadvantages of Virtual Machines

* Performance is slower than physical machines
* High RAM/CPU usage
* Requires powerful hardware for multiple VMs

---

Image is a **blueprint**.

It contains instructions like:

* Copy app code
* Install libraries

It’s just instructions.

```
Image contains:
- Python 3.11
- Your app
- Required libraries
```

Think of image as a template.

# 2. Docker Container = Running App 📦

Container is the **actual running instance of the image**.

Image → blueprint
Container → running software

You can create many containers from one image.

Example:

```
Image: myapp

Containers:
Container 1 → running instance
Container 2 → running instance
Container 3 → running instance
```

Analogy:

Recipe → Image
Cake → Container

You can bake many cakes from one recipe.

---

# 3. Docker Engine = Kitchen ⚙️

Docker Engine is the software that runs containers.

It:

* Creates containers
* Runs containers
* Stops containers
* Manages containers

Without Docker Engine, containers cannot run.

### Interview questions 

1.How do you debug a running container?
```
docker ps
docker logs <container_id>
docker exec -it <container_id> /bin/sh
```

2.How do you reduce Docker image size?

- Use a suitable lightweight base image.
- Install only required dependencies.
- Use .dockerignore. (.dockerignore is used to ignore file to be considered in image)
- Avoid copying training datasets and unnecessary files.
- Remove package-manager caches where appropriate.
- Use multi-stage builds when they provide a benefit.
